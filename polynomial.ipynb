```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("polynominal_dataset.csv")
df
```

```python
# step 3:  EDA

plt.scatter(df['Ice Cream Sales (units)'],df['Temperature (°C)'])
plt.title('plot the data')
plt.xlabel("temperature")
plt.ylabel("Ice Cream Sales(units)")
plt.show()
```

```python
# step5; Divide the into x and y
x = df['Temperature (°C)'].value.reshape(-1,1)
y =df['Ice Cream Sales (units)']
```

```python
# Polynomial Regression
from sklearn.preprocessing import PolynomialFeatures

# Convert x from 1D Series to 2D
x = x.to_frame()

# Create the instance
poly = PolynomialFeatures(degree=2)

# Fit and transform
x_poly = poly.fit_transform(x)

x_poly
```

```python
# step 6: Model building
from sklearn.linear_model import LinearRegression

# create the instance of model
LR = LinearRegression()
#Train the model on the the x_poly
model = LR.fit(x_poly,y)

#model predictions
y_pred = model.predict(x_poly)
```

```python
# model evaluation metrices
 
from sklearn.metrics import mean_absolute_error, mean_squared_error,r2_score

#find MSE
MSE = mean_squared_error(y,y_pred)

#find MAE
MAE = mean_absolute_error(y,y_pred)

# find the R2_socre
r2_score = r2_score(y,y_pred)

#print all the metrices
print(f"MSE: ")
```
